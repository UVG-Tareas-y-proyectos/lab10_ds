## Inspección de la cobertura y los agregados
Las sumas por región, mercado y frontera deben conservar el total mensual. La cobertura de 2026 llega a junio.

```python
import pandas as pd
import numpy as np
D = ROOT/'data/processed'
m = pd.read_csv(D/'serie_mensual_total.csv',parse_dates=['fecha'])
for nombre in ['mercados','regiones','fronteras']:
    g = pd.read_csv(D/f'{nombre}.csv').groupby('fecha').viajeros.sum()
    np.testing.assert_allclose(g,m.viajeros)
m.tail(12)
```
